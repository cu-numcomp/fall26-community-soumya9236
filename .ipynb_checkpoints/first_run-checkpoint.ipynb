```julia
using Pkg
Pkg.add("IterativeSolvers")
```

```julia
using IterativeSolvers
```

```julia
## EXAMPLE 

using IterativeSolvers
using LinearAlgebra

A = [4.0 1.0; 1.0 3.0]
b = [1.0, 2.0]

x = cg(A, b)

println("Solution using Congugate Gradient: ", x)
## comparing with direct solver 
x_direct = A \ b
println("Solution using direct solver: ", x_direct)
#residual
residual = norm(A * x - b)
println("Residual: ", residual)
```

```julia
# EXAMPLE CHANGED
using IterativeSolvers
using LinearAlgebra
A = [4.0 1.0 0.0;
     1.0 3.0 1.0;
     0.0 1.0 2.0]
b = [1.0, 2.0, 3.0]
x, history = cg(A, b, log=true)
println("Solution: ", x)
println("Number of iterations: ", history.iters)
println("Converged: ", history.isconverged)
println("Residual: ", norm(A*x - b))
```
